```python
# Import vizualization packages
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from matplotlib.ticker import PercentFormatter
from matplotlib.cbook import boxplot_stats
from IPython.display import display

# Formatter to display y-axis numbers with commas
formatter = FuncFormatter(lambda x, _: f'{int(x):,}')

# Group data by claim status
claim_status_group = tiktok_videos_df.groupby(['claim_status'])
```

```python
def describe_claim_status_by_engagement():
    # Define a custom aggregation function with named output
    agg_funcs = {
        'video_view_count': ['mean',
                             'median',
                             lambda x: x.quantile(0.25),
                             lambda x: x.quantile(0.75),
                             lambda x: x.quantile(0.75) - x.quantile(0.25)],
        'video_like_count': ['mean',
                             'median',
                             lambda x: x.quantile(0.25),
                             lambda x: x.quantile(0.75),
                             lambda x: x.quantile(0.75) - x.quantile(0.25)],
        'video_comment_count': ['mean',
                                'median',
                                lambda x: x.quantile(0.25),
                                lambda x: x.quantile(0.75),
                                lambda x: x.quantile(0.75) - x.quantile(0.25)],
        'video_share_count': ['mean',
                              'median',
                              lambda x: x.quantile(0.25),
                              lambda x: x.quantile(0.75),
                              lambda x: x.quantile(0.75) - x.quantile(0.25)]
    }

    # Perform the aggregation
    grouped = tiktok_videos_df.groupby('claim_status').agg(agg_funcs)

    # Flatten MultiIndex columns to strings
    grouped.columns = ['_'.join(col).strip() for col in grouped.columns.values]

    # Rename the lambda-generated columns
    grouped.columns = [col.replace('<lambda_0>', 'Q1')
                          .replace('<lambda_1>', 'Q3')
                          .replace('<lambda_2>', 'IQR')
                          .replace('video_view_count', 'view')
                          .replace('video_like_count', 'like')
                          .replace('video_share_count', 'share')
                          .replace('video_comment_count', 'comment')
                       for col in grouped.columns]
    # Reset index if needed
    grouped = grouped.reset_index()
    pd.set_option('display.max_columns', None)
    return grouped

def describe_claim_status_by_column(column_name):
    # Define the aggregation functions
    agg_funcs = {
        column_name: [lambda x: round(x.mean(), 2),
                      # 'mean',
                      'median',
                      lambda x: x.quantile(0.25),
                      lambda x: x.quantile(0.75),
                      lambda x: x.quantile(0.75) - x.quantile(0.25),
                      'min', 'max']
    }

    # Perform the aggregation
    grouped = tiktok_videos_df.groupby('claim_status').agg(agg_funcs)

    # Flatten and clean column names
    # prefix = column_name.replace('video_', '').replace('_count', '')
    # grouped.columns = [f'{prefix}_mean', f'{prefix}_median', f'{prefix}_Q1', f'{prefix}_Q3', f'{prefix}_IQR']
    grouped.columns = ['mean', 'median', 'Q1', 'Q3', 'IQR', 'min', 'max']

    # Reset index for clean output
    grouped = grouped.reset_index()

    pd.set_option('display.max_columns', None)
    return grouped

def plot_bar_by_claim_author_status():
    # Grouped data
    grouped_data = tiktok_videos_df.groupby(['claim_status', 'author_ban_status']).size().reset_index(name='count')

    plt.figure(figsize=(10, 6))
    barplot = sns.barplot(data=grouped_data, x='claim_status', y='count', hue='author_ban_status')

    # Manually add text labels
    for i in range(len(grouped_data)):
        claim = grouped_data['claim_status'][i]
        author = grouped_data['author_ban_status'][i]
        count = grouped_data['count'][i]

        # Get x position of the bar
        x = list(grouped_data['claim_status'].unique()).index(claim)
        hue_offset = ['active', 'banned', 'under review'].index(author) * 0.2 - 0.2  # adjust bar spacing
        plt.text(x + hue_offset, count + 100, str(count), ha='center', va='bottom', fontsize=9)

    plt.title('Video Count by Claim Status and Author Ban Status')
    plt.xlabel('Claim Status')
    plt.ylabel('Number of Videos')
    plt.legend(title='Author Ban Status')
    plt.tight_layout()
    plt.show()
        
def plot_box_by_claim_status_engagement(column_name, title, ax, show_legend=False):
    sns.boxplot(data=tiktok_videos_df, x='claim_status', y=column_name, ax=ax)
    ax.yaxis.set_major_formatter(formatter)
    ax.set_title(f'{title} by Claim Status')
    ax.set_xlabel('Claim Status')
    ax.set_ylabel(f'{title} Count')

    # Plot means
    means = tiktok_videos_df.groupby('claim_status')[column_name].mean()
    for i, mean in enumerate(means):
        ax.scatter(i, mean, color='red', marker='D', s=40, zorder=5, label='Mean' if i == 0 else "")
        ax.hlines(mean, xmin=i - 0.4, xmax=i + 0.4, colors='red', linestyles='dashed', linewidth=1)
        ax.text(i, mean, f'{int(mean):,}', color='red', ha='center', va='bottom', fontsize=8)

    # Calculate boxplot stats
    for i, (name, group) in enumerate(tiktok_videos_df.groupby('claim_status')):
        stats = boxplot_stats(group[column_name])[0]  # get stats for this group

        # Min and Max (whiskers)
        ax.text(i - 0.15,
                stats['whislo'],
                f'Lower Whisker:\n{int(stats["whislo"]):,}',
                color='gray',
                ha='right',
                va='center',
                fontsize=8)
        ax.text(i - 0.15,
                stats['whishi'],
                f'Upper Whisker:\n{int(stats["whishi"]):,}',
                color='gray',
                ha='right',
                va='center',
                fontsize=8)

        # Q1 and Q3
        ax.text(i + 0.15, stats['q1'], f'Q1:\n{int(stats["q1"]):,}', ha='left', va='center', fontsize=8)
        ax.text(i + 0.15, stats['q3'], f'Q3:\n{int(stats["q3"]):,}', ha='left', va='center', fontsize=8)

    if show_legend:
        ax.legend()

def plot_box_by_claim_status():
    fig, axes = plt.subplots(4, 1, figsize=(10, 24))

    # Plot each engagement metric
    plot_box_by_claim_status_engagement('video_view_count', 'Views', axes[0], show_legend=True)
    plot_box_by_claim_status_engagement('video_like_count', 'Likes', axes[1])
    plot_box_by_claim_status_engagement('video_share_count', 'Shares', axes[2])
    plot_box_by_claim_status_engagement('video_comment_count', 'Comments', axes[3])

    plt.tight_layout()
    plt.show()

def plot_single_box_by_claim_status(column_name, title, show_legend=False):
    fig, ax = plt.subplots(figsize=(10, 6))
    plot_box_by_claim_status_engagement(column_name, title, ax, show_legend=show_legend)
    plt.tight_layout()
    plt.show()
    print(title)
    display(describe_claim_status_by_column(column_name))

def plot_histogram_by_engagement(engagement_col, engagement_label):
    plt.figure(figsize=(12, 10))
    
    video_count_label = 'Number of Videos'

    plt.subplot(2, 1, 1)
    hist = sns.histplot(tiktok_videos_df[engagement_col], 
                        bins=50,
                        kde=True)
    plt.title(f'Distribution of {engagement_label} Overall')
    plt.xlabel(engagement_label)
    plt.ylabel(video_count_label)
    
    # Format x-axis as percentage
    plt.gca().xaxis.set_major_formatter(PercentFormatter(xmax=1))
    
    # Rotate x-axis labels
    # plt.xticks(rotation=45)
    
    # Add value labels on top of bars
    for patch in hist.patches:
        height = patch.get_height()
        if height > 0:
            plt.text(patch.get_x() + patch.get_width() / 2,
                     height + 5,
                     f'{int(height)}',
                     ha='center',
                     va='bottom',
                     fontsize=8,
                     rotation=45)

    # Plot histogram for each claim_status
    plt.subplot(2, 1, 2)
    hist_by_claim = sns.histplot(data=tiktok_videos_df,
                                 x=engagement_col, 
                                 hue='claim_status',
                                 bins=50,
                                 kde=True,
                                 element='bars', # 'step'
                                 stat='count',
                                 common_norm=False)
    plt.title(f'Distribution of {engagement_label} by Claim Status')
    plt.xlabel(engagement_label)
    plt.ylabel(video_count_label)
    plt.gca().xaxis.set_major_formatter(PercentFormatter(xmax=1))
    # plt.xticks(rotation=45)
    
    # Add value labels per hue group
    for container in hist_by_claim.containers:
        for bar in container:
            height = bar.get_height()
            if height > 0:
                hist_by_claim.annotate(f'{int(height)}',
                                       xy=(bar.get_x() + bar.get_width() / 2, height),
                                       xytext=(0, 4),
                                       textcoords='offset points',
                                       ha='center',
                                       va='bottom',
                                       fontsize=8,
                                       rotation=45)

    plt.tight_layout()
    plt.show()
    
    # Show metrics
    likes_per_view_summary_df = pd.DataFrame([{
        'count': tiktok_videos_df[engagement_col].count(),
        'mean': tiktok_videos_df[engagement_col].mean(),
        'median': tiktok_videos_df[engagement_col].median(),
        'min': tiktok_videos_df[engagement_col].min(),
        'max': tiktok_videos_df[engagement_col].max()
    }])
    print(f'Overall {engagement_label}')
    display(likes_per_view_summary_df)
    display(claim_status_group.agg({engagement_col: ['count', 'mean', 'median', 'min', 'max']}))
    
def plot_heatmap_by_engagement():
    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.figure(figsize=(10, 8))

    # Original column names
    numeric_cols = ['video_view_count',
                    'video_like_count',
                    'video_share_count',
                    'video_download_count',
                    'video_comment_count']

    # Custom labels
    label_map = {
        'video_view_count': 'Views',
        'video_like_count': 'Likes',
        'video_share_count': 'Shares',
        'video_download_count': 'Downloads',
        'video_comment_count': 'Comments'
    }

    # Compute correlation and rename for display
    correlation = tiktok_videos_df[numeric_cols].corr()
    correlation.rename(index=label_map, columns=label_map, inplace=True)

    sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Correlation Matrix of Engagement Metrics')
    plt.show()
```

### 📌 Data Analyst Notes

#### Retrieving Q1, Q3, IQR, Lower Whisker, and Upper Whisker

**Given observations:** `[10, 20, 30, 40, 50]`  
`n = 5` (number of observations)  
`min = 10`  
`max = 50`  

---

**Quartile Formula:**  
Q = value_at_floor + (position - floor) * (value_at_ceil - value_at_floor)

---

#### 🔹 Q1: 25th percentile (Lower Quartile)
    Q1 position = (n + 1) * 0.25
                = (5 + 1) * 0.25
                = 1.5

    floor index           = 1
    ceil index            = 2
    value at floor index  = 10
    value at ceil index   = 20

    Q1 = 10 + (1.5 - 1) * (20 - 10)
       = 10 + 0.5 * 10
       = 15

#### 🔹 Q3: 75th percentile (Upper Quartile)
    Q3 position = (n + 1) * 0.75
                = (5 + 1) * 0.75
                = 4.5

    floor index           = 4
    ceil index            = 5
    value at floor index  = 40
    value at ceil index   = 50

    Q3 = 40 + (4.5 - 4) * (50 - 40)
       = 40 + 0.5 * 10
       = 45

---

#### 🔸 IQR: Interquartile Range
    IQR = Q3 - Q1
        = 45 - 15
        = 30

---

#### 🔸 Whiskers
   ##### Upper whisker:  
    Lower whisker = Q1 - 1.5 * IQR  
                  = 15 - 1.5 * 30  
                  = 15 - 45  
                  = -30  

        Actual minimum = 10  
        Since 10 ≥ -30 → Lower whisker is 10

   ##### Upper whisker:  
    Upper whisker = Q3 + 1.5 * IQR  
                  = 45 + 1.5 * 30  
                  = 45 + 45  
                  = 90  

        Actual maximum = 50  
        Since 50 ≤ 90 → Upper whisker is 50
